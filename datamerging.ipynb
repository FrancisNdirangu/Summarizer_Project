```python
import pandas as pd 
```

```python
df_ai = pd.read_csv('data/ai_ml_articles_20250723_132012.csv')
df_ai = df_ai.drop(columns=['url'], axis = 0)
df_ai.head()
```

```python
df_ar = pd.read_csv('data/arxiv_ai_abstracts.csv')
df_ar['content'] = df_ar['Content']
df_ar['title'] = df_ar['Title']
df_ar = df_ar.drop(columns=['Content', 'Title'], axis = 0)
df_ar['word_count'] = df_ar['content'].apply(lambda x: len(x.split()))
df_ar.head()
```

```python
df_blog = pd.read_csv('data/blogspot_ai_ml_scraped_data_100_per_blog.csv')
df_blog = df_blog.drop(columns=['url'], axis=0)
df_blog['word_count'] = df_blog['content'].apply(lambda x: len(x.split()))
df_blog.head()
```

```python
df_ds = pd.read_csv('data/datascienceplus_articles.csv')
df_ds['title'] = df_ds['Title']
df_ds['word_count'] = df_ds['Content'].apply(lambda x: len(str(x).split()) if pd.notnull(x) else 0)
df_ds['content'] = df_ds['Content']
df_ds = df_ds.drop(columns=['URL', 'Date', 'Content', 'Title'], axis=0)
df_ds.head()
```

```python

df_last = pd.read_csv('data/last20250725_122248.csv')
df_last = df_last.drop(columns=['url'], axis = 0)
df_last.head()
```

```python
df_data = pd.read_csv('data/Ltr20250724_123704.csv')
df_data_ = df_data.drop(columns=['url'], axis=0, inplace=True)
df_data.head()
```

```python
df_sh = pd.read_csv('data/shyambhu_all_posts.csv')
df_sh['content'] = df_sh['text']
df_sh = df_sh.drop(columns=['url', 'published_date', 'text'], axis=0)
df_sh['word_count'] = df_sh['content'].apply(lambda x: len(x.split()))
df_sh.head()
```

```python
df_1 = pd.read_csv('data/ian1.csv')
df_1 = df_1.drop(columns=['url'])
df_1.head()
```

```python
df2 = pd.read_csv('data/ian2.csv')
df2 = df2.drop(columns=['url'])
df2.head()
```

```python
df3 = pd.read_csv('data/ian3.csv')
df3 = df3.drop(columns=['url'])
df3.head()
```

```python
df_merge = pd.concat([df_ai, df_ar, df_blog, df_ds, df_last, df_data, df_sh, df_1, df2, df3], ignore_index=True)
df_merge.head()
```

```python
df_merge.shape
```

```python
df_merge.duplicated().value_counts()
```

```python
df_merge = df_merge.drop_duplicates(subset=['title', 'content'], keep='first')
df_merge.shape
```

```python
df_merge = df_merge[df_merge['word_count'] >= 100]
df_merge.shape
```

```python
df_merge.to_csv('data/merged_data.csv', index=False)
print("Data merging completed and saved to 'data/merged_data.csv'.")  
```
