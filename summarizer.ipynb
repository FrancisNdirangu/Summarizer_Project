```python
import pandas as pd
import os
```

```python
directory = r"C:\Users\HP\OneDrive\Desktop\summarizer"

# List of CSV files to merge
csv_files = [
    "Ltr20250724_122235.csv",
    "Ltr20250724_123704.csv",
    "later 220250724_121426.csv",  
    "ai_ml_articles_20250723_132012.csv",
    "arxiv_ai_abstracts.csv",
    "datascienceplus_articles.csv",
    "last20250725_122248.csv",
    "later20250724_120426.csv",
    "shyambhu_all_posts.csv"
]

# Initialize an empty list to store DataFrames
dataframes = []

# Load each CSV file
for file in csv_files:
    file_path = os.path.join(directory, file)
    if os.path.exists(file_path):  
        try:
            df = pd.read_csv(file_path)
            dataframes.append(df)
            print(f"Successfully loaded: {file}")
        except Exception as e:
            print(f"Error loading {file}: {e}")
    else:
        print(f"File not found: {file}")

# Merge all DataFrames
if dataframes:
    merged_data = pd.concat(dataframes, ignore_index=True)
    
    # Optional: Remove duplicates based on 'title' and 'url' if they exist
    if all(col in merged_data.columns for col in ['title', 'url']):
        merged_data = merged_data.drop_duplicates(subset=['title', 'url'], keep='first')
    
    # Save the merged DataFrame to a new CSV
    output_file = os.path.join(directory, "merged_data.csv")
    merged_data.to_csv(output_file, index=False, encoding='utf-8')
    print(f"Merged {len(merged_data)} articles and saved to {output_file}")
else:
    print("No files were loaded successfully.")

# Display the first few rows of the merged DataFrame 
if 'merged_data' in locals():
    print("\nPreview of merged data:")
    print(merged_data.head())
```

```python
cols_to_drop = ['Title', 'Content', 'URL', 'Date', 'published_date', 'text']
merged_data = merged_data.drop(columns=[col for col in cols_to_drop if col in merged_data.columns])

# Drop rows where content is missing
merged_data = merged_data.dropna(subset=['content'])

# Reset index
merged_data = merged_data.reset_index(drop=True)

print("Cleaned shape:", merged_data.shape)
merged_data.head()
```

```python
from transformers import pipeline

# Load summarization pipeline with distilBART
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

# Example long text
text = """Machine learning is a subset of artificial intelligence that focuses on building systems that learn from and make decisions based on data. These systems improve over time without being explicitly programmed."""

# Get summary
summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
print("Summary:", summary[0]['summary_text'])
```
